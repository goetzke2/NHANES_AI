```python
import pandas as pd
import numpy as np
from sklearn import metrics
import matplotlib.pyplot as plt
```

#### NHANES

```python
#import dataframe output from table 1
#EyeFields3 file edited to fill ungradeable to 3's
df = pd.read_csv(r"Z:\Output\Data tables\EyeFields3.csv")
df = df[['SEQN','AMD']]
```

```python
print('N of Respondents: ', df['SEQN'].nunique())
df.shape
```

```python
df.isna().sum()
```

```python
df['AMD'].value_counts()
```

#### Model A

```python
#import A_output
a = pd.read_csv(r"Z:\Data\A_output_upd.csv")
a = a[['SEQN','Laterality', 'AMD']]
a = a.pivot(index = 'SEQN', columns = 'Laterality', values = 'AMD')
a = pd.merge(df, a, on='SEQN')
print('N in NHANES and Model A: ', a['SEQN'].nunique())
```

```python
a['Worse'] = a[['OD', 'OS']].max(axis=1)
a = a[['SEQN','AMD', 'Worse']]
```

```python
a.isna().sum()
```

```python
a = a.dropna()
```

```python
a['SEQN'].nunique()
```

```python
actual = a['AMD']
predicted = a['Worse']

confusion_matrix = metrics.confusion_matrix(actual, predicted)
```

```python
cm_display = metrics.ConfusionMatrixDisplay(confusion_matrix = confusion_matrix)
cm_display.plot()
plt.show()
```

```python
 from sklearn.metrics import classification_report
 print(classification_report(actual, predicted))
```

#### ROC 
*probs for models b, g, h

```python
from sklearn.metrics import roc_curve
from sklearn.metrics import roc_auc_score
from matplotlib import pyplot
```

#### No AMD

```python
prob_b = pd.read_csv(r"Z:\Data\probs-B_upd.csv")
prob_g = pd.read_csv(r"Z:\Data\prob - G_upd.csv")
prob_h = pd.read_csv(r"Z:\Data\prob-H_upd.csv")

prob_b = prob_b[['SEQN','Laterality', 'No_AMD_prob']]
prob_b = prob_b.pivot(index = 'SEQN', columns = 'Laterality', values = 'No_AMD_prob')
prob_b = pd.merge(df, prob_b, on='SEQN')

prob_g = prob_g[['SEQN','Laterality', 'No_AMD_prob']]
prob_g = prob_g.pivot(index = 'SEQN', columns = 'Laterality', values = 'No_AMD_prob')
prob_g = pd.merge(df, prob_g, on='SEQN')

prob_h = prob_h[['SEQN','Laterality', 'No_AMD_prob']]
prob_h = prob_h.pivot(index = 'SEQN', columns = 'Laterality', values = 'No_AMD_prob')
prob_h = pd.merge(df, prob_h, on='SEQN')
```

```python
prob_b['Worse'] = prob_b[['OD', 'OS']].max(axis=1)
prob_b = prob_b[['SEQN','AMD', 'Worse']]

prob_g['Worse'] = prob_g[['OD', 'OS']].max(axis=1)
prob_g = prob_g[['SEQN','AMD', 'Worse']]

prob_h['Worse'] = prob_h[['OD', 'OS']].max(axis=1)
prob_h = prob_h[['SEQN','AMD', 'Worse']]
```

```python
prob_b = prob_b.dropna()
prob_g = prob_g.dropna()
prob_h = prob_h.dropna()

print ("N_B: ", prob_b['SEQN'].nunique())
print ("N_G: ", prob_g['SEQN'].nunique())
print ("N_H: ", prob_h['SEQN'].nunique())
```

```python
actual = prob_b['AMD']
predicted = prob_b['Worse']
b_fpr, b_tpr, thresholds = metrics.roc_curve(actual, predicted, pos_label=0)

actual = prob_g['AMD']
predicted = prob_g['Worse']
g_fpr, g_tpr, thresholds = metrics.roc_curve(actual, predicted, pos_label=0)

actual = prob_h['AMD']
predicted = prob_h['Worse']
h_fpr, h_tpr, thresholds = metrics.roc_curve(actual, predicted, pos_label=0)
```

```python
pyplot.plot(b_fpr, b_tpr, label='Model B')
pyplot.plot(g_fpr, g_tpr, label='Model G')
pyplot.plot(h_fpr, h_tpr, label='Model H')
pyplot.xlabel('False Positive Rate')
pyplot.ylabel('True Positive Rate')
pyplot.legend()
pyplot.show()
```

```python
plt.savefig(r"Z:\Output\Data tables\ROC\No_AMD.svg", format='svg', dpi=1200)
```
